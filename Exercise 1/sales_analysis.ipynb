# Sales Analysis

This is the sales data for the first 6 months of the year. I'm using pandas to add up the numbers, see how sales changed over time, and then give a few ideas on how the company could do better.

## Dataset

```python
import pandas as pd

sales = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "revenue": [80000, 85000, 82000, 95000, 110000, 130000],
    "products_sold": [400, 420, 410, 450, 500, 550]
})

sales
```

## Part A - Data Analysis

```python
# 1- Total sales revenue for the 6 months
total_revenue = sales["revenue"].sum()
print("Total revenue ($):", total_revenue)
```

```python
# 2- Average monthly revenue
average_revenue = sales["revenue"].mean()
print("Average monthly revenue ($):", average_revenue)
```

```python
# 3- Month with the highest revenue
highest_index = sales["revenue"].idxmax()
highest_month = sales.loc[highest_index, "month"]
print("Highest revenue month:", highest_month, "-", sales.loc[highest_index, "revenue"])
```

```python
# 4- Month with the lowest revenue
lowest_index = sales["revenue"].idxmin()
lowest_month = sales.loc[lowest_index, "month"]
print("Lowest revenue month:", lowest_month, "-", sales.loc[lowest_index, "revenue"])
```

```python
# 5- Total number of products sold
total_products = sales["products_sold"].sum()
print("Total products sold:", total_products)
```

```python
# 6- Average revenue per product for each month
sales["revenue_per_product"] = (sales["revenue"] / sales["products_sold"]).round(2)
print(sales[["month", "revenue_per_product"]].to_string(index=False))
```

```python
# 7- Percentage increase in revenue from Jan to Jun
jan_revenue = sales.loc[0, "revenue"]
jun_revenue = sales.loc[5, "revenue"]
revenue_increase = ((jun_revenue - jan_revenue) / jan_revenue * 100).round(2)
print("Revenue increase Jan to Jun (%):", revenue_increase)
```

## Part B - Identifying Trends

```python
# Work out the change from the month before
sales["change"] = sales["revenue"].diff()
sales["change_pct"] = (sales["revenue"].pct_change() * 100).round(2)
sales[["month", "revenue", "change", "change_pct"]]
```

```python
# 1- Did revenue go up every month?
# is_monotonic_increasing is only True if each month is higher than the one before
print("Revenue increased every month:", sales["revenue"].is_monotonic_increasing)
```

```python
# 2- Which month had a drop in revenue compared to the month before?
drop_months = sales[sales["change"] < 0]
print(drop_months[["month", "change"]].to_string(index=False))
```

```python
# 3- Change between Feb and Mar ($ and %)
feb_revenue = sales.loc[1, "revenue"]
mar_revenue = sales.loc[2, "revenue"]
feb_mar_change = mar_revenue - feb_revenue
feb_mar_pct = (feb_mar_change / feb_revenue * 100).round(2)
print("Change Feb to Mar ($):", feb_mar_change)
print("Change Feb to Mar (%):", feb_mar_pct)
```

```python
# 4- Did products sold go up from Jan to Jun?
jan_products = sales.loc[0, "products_sold"]
jun_products = sales.loc[5, "products_sold"]
products_increase = ((jun_products - jan_products) / jan_products * 100).round(2)
print("Products sold Jan:", jan_products)
print("Products sold Jun:", jun_products)
print("Increase (%):", products_increase)
```

```python
# 5- Did average revenue per product go up from Jan to Jun?
jan_rpp = sales.loc[0, "revenue_per_product"]
jun_rpp = sales.loc[5, "revenue_per_product"]
rpp_increase = ((jun_rpp - jan_rpp) / jan_rpp * 100).round(2)
print("Revenue per product Jan ($):", jan_rpp)
print("Revenue per product Jun ($):", jun_rpp)
print("Increase (%):", rpp_increase)
```

## Part C - Business Interpretation

```python
# Did revenue go up because we sold more, or because we made more on each product?
# revenue = products sold x revenue per product, so I can check each one on its own
volume_growth = jun_products / jan_products
price_growth = (jun_revenue / jun_products) / (jan_revenue / jan_products)
print("Volume growth:", round(volume_growth, 3))
print("Revenue per product growth:", round(price_growth, 3))
print("Together:", round(volume_growth * price_growth, 3))
```

```python
# Three key findings
# 1- Revenue went from $80,000 in Jan to $130,000 in Jun, which is a 62.5%
#    increase. Total revenue for the 6 months was $582,000.
# 2- Revenue dropped in Mar, it was $3,000 lower than Feb. Every other
#    month went up.
# 3- Products sold went up (400 to 550) and so did revenue per product
#    ($200 to $236.36) between Jan and Jun.

# Is the growth from more sales, higher revenue per product, or both?
# Both. Products sold went up 37.5% and revenue per product went up 18.18%,
# and together that gives the 62.5%. More products sold had the bigger effect.

# Two recommendations
# 1- Check what was different from Apr to Jun and keep doing it. Also make
#    sure there is enough stock for the higher sales.
# 2- Check why Mar dropped and run promotions in the slower months.
```
